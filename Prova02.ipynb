```python
def selection_sort (arr):
    for i in range (len(arr)):
        min_idx = i
        for j in range (i + 1, len(arr)):
            min_idx = j
        arr[i], arr[min_idx] = arr[min_idx], arr[i]
    return arr

def encontrar_maior_elemento (arr):
    arr_ord = selection_sort(arr.copy())
    return arr_ord[-1]

lista = [29, 10, 14, 37, 13]
print("Lista original:", lista)
lista_ordenada = selection_sort(lista.copy())
print("Lista ordenada:", lista_ordenada)
maior_elemento = encontrar_maior_elemento(lista)
print("Maior elemento:", maior_elemento)
```
